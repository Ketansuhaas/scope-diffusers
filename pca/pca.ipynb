```python
import torch
from sklearn.decomposition import PCA
import torch.nn.functional as F

# Generate random tensors of shape (77, 1024)
tensor1 = torch.randn(77, 1024)
tensor2 = torch.randn(77, 1024)

# Convert to numpy for PCA (ensure tensors are on CPU)
matrix1 = tensor1.cpu().numpy()
matrix2 = tensor2.cpu().numpy()

# Perform PCA on tensor1
n_components = 50  # Adjust based on variance retained
pca1 = PCA(n_components=n_components)
pca1.fit(matrix1)

# Transform tensor2 into the subspace of tensor1
matrix2_transformed = pca1.transform(matrix2)

# Reconstruct tensor2 in the subspace of tensor1
matrix2_reconstructed = pca1.inverse_transform(matrix2_transformed)

# Convert back to PyTorch tensor efficiently
tensor2_reconstructed = torch.from_numpy(matrix2_reconstructed)

print("Original Tensor2 Shape:", tensor2.shape)
print("Reconstructed Tensor2 Shape:", tensor2_reconstructed.shape)

# Compute cosine similarities
cos_sim_t1_t2 = F.cosine_similarity(tensor1, tensor2).mean()
cos_sim_t1_t2_recon = F.cosine_similarity(tensor1, tensor2_reconstructed).mean()
cos_sim_t2_t2_recon = F.cosine_similarity(tensor2, tensor2_reconstructed).mean()

# Compute Mean Squared Errors (MSE)
mse_t1_t2 = F.mse_loss(tensor1, tensor2)
mse_t1_t2_recon = F.mse_loss(tensor1, tensor2_reconstructed)
mse_t2_t2_recon = F.mse_loss(tensor2, tensor2_reconstructed)

# Print similarity metrics
print("\nCosine Similarities:")
print(f"Tensor1 vs Tensor2: {cos_sim_t1_t2.item():.4f}")
print(f"Tensor1 vs Tensor2_Recon: {cos_sim_t1_t2_recon.item():.4f}")
print(f"Tensor2 vs Tensor2_Recon: {cos_sim_t2_t2_recon.item():.4f}")

print("\nMean Squared Errors (MSE):")
print(f"Tensor1 vs Tensor2: {mse_t1_t2.item():.4f}")
print(f"Tensor1 vs Tensor2_Recon: {mse_t1_t2_recon.item():.4f}")
print(f"Tensor2 vs Tensor2_Recon: {mse_t2_t2_recon.item():.4f}")
```
